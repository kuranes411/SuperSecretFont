```python
import tkinter as tk
from tkinter import Canvas
import time

glyph_faded = False 

#window canvas
root = tk.Tk()
root.title("Living Glyph")
root.geometry("400x400")
root.configure(bg='white')

canvas = Canvas(root, width=400, height=400, bg='white', highlightthickness=0)
canvas.pack()


#points for glyph
glyph = canvas.create_polygon(
200, 50,  
150, 350,  
170, 350,  
200, 150,  
230, 350,
250, 350,  
fill='black'
)

#fading
def fade_out():
    global glyph_faded
    if glyph_faded:
        return  
    steps = 20
    for i in range(steps):
        gray_value = int(255 * (i+1) / steps)
        color = f'#{gray_value:02x}{gray_value:02x}{gray_value:02x}' 
        canvas.itemconfig(glyph, fill=color)
        root.update()
        time.sleep(0.05)
    glyph_faded = True  


#clipboard c
previous_clipboard = None

def check_clipboard():
    global previous_clipboard
    try:
        current_clipboard = root.clipboard_get()
        if previous_clipboard != current_clipboard:
            print("Clipboard changed (text) — OK")
            previous_clipboard = current_clipboard
    except tk.TclError:
        print("Clipboard changed (non-text — maybe an image?)")
        fade_out()
    root.after(500, check_clipboard) 

root.after(500, check_clipboard)

root.mainloop()
```

```python
import tkinter as tk
from tkinter import Canvas, simpledialog
import random
import time

# canvas window
root = tk.Tk()
root.title("OCR-Confusing Glyph")
root.geometry("400x400")
root.configure(bg='white')

canvas = Canvas(root, width=400, height=400, bg='white', highlightthickness=0)
canvas.pack()

#prompt
char = simpledialog.askstring("Input", "Enter a Unicode character:")

if not char:
    char = "A"  # default

#glyph
glyph = canvas.create_text(
    200, 200,
    text=char,
    font=("Helvetica", 160, "bold"),
    fill='black'
)

#dots to confuse OCR
for _ in range(50):
    nx = random.randint(120, 280)
    ny = random.randint(120, 280)
    size = random.randint(1, 2)
    canvas.create_oval(nx, ny, nx + size, ny + size, fill='black', outline='')


#did it fade
glyph_faded = False

#fading
def fade_out():
    global glyph_faded
    if glyph_faded:
        return
    steps = 20
    for i in range(steps):
        gray_value = int(255 * (i+1) / steps)
        color = f'#{gray_value:02x}{gray_value:02x}{gray_value:02x}'
        canvas.itemconfig(glyph, fill=color)
        root.update()
        time.sleep(0.05)
    glyph_faded = True
    
#clipboard
previous_clipboard = None

def check_clipboard():
    global previous_clipboard
    try:
        current_clipboard = root.clipboard_get()
        if previous_clipboard != current_clipboard:
            print("Clipboard changed (text) — OK")
            previous_clipboard = current_clipboard
    except tk.TclError:
        print("Clipboard changed (non-text — maybe an image?)")
        fade_out()
    root.after(500, check_clipboard)

root.after(500, check_clipboard)

root.mainloop()
```
